```python
df = df.drop(columns=["customerID"])
```

```python
import matplotlib.pyplot as plt
contract_churn = pd.crosstab(df["Contract"], df["Churn"], normalize="index")
contract_churn
```

```python
contract_churn.plot(kind="bar")

plt.title("Churn Rate by Contract Type")
plt.xlabel("Contract Type")
plt.ylabel("Proportion")
plt.xticks(rotation=0)

plt.savefig(
    "../results/churn_by_contract.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

Customers on month-to-month contracts show a substantially higher churn rate than customers on one-year or two-year contracts

```python
df.groupby("Churn")["tenure"].mean()
```

```python
df.boxplot(column="tenure", by="Churn")

plt.title("Tenure by Churn")
plt.suptitle("")
plt.xlabel("Churn")
plt.ylabel("Tenure (Months)")

plt.savefig(
    "../results/churn_by_tenure.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
df["tenure_group"] = pd.cut(
    df["tenure"],
    bins=[0, 12, 24, 48, 72],
    labels=["0-12", "13-24", "25-48", "49-72"],
    include_lowest=True
)
```

```python
tenure_churn = pd.crosstab(
    df["tenure_group"],
    df["Churn"],
    normalize="index"
)

tenure_churn
```

```python
tenure_churn.plot(kind="bar")

plt.title("Churn Rate by Tenure Group")
plt.xlabel("Tenure Group (Months)")
plt.ylabel("Proportion")
plt.xticks(rotation=0)
plt.show()
```

Customers with shorter tenure show substantially higher churn rates, while long-term customers are less likely to churn.

```python
df.groupby("Churn")["MonthlyCharges"].mean()
```

```python
df.boxplot(column="MonthlyCharges", by="Churn")

plt.title("Monthly Charges by Churn")
plt.suptitle("")
plt.xlabel("Churn")
plt.ylabel("Monthly Charges")
plt.show()
```

```python
df["monthly_charge_group"] = pd.cut(
    df["MonthlyCharges"],
    bins=[0, 40, 70, 100, 120],
    labels=["0-40", "41-70", "71-100", "101-120"],
    include_lowest=True
)
```

```python
charge_churn = pd.crosstab(
    df["monthly_charge_group"],
    df["Churn"],
    normalize="index"
)

charge_churn

charge_churn.plot(kind="bar")

plt.title("Churn Rate by Monthly Charges")
plt.xlabel("Monthly Charges")
plt.ylabel("Proportion")
plt.xticks(rotation=0)
plt.show()
```

Customers with higher monthly charges tend to show higher churn rates.

```python
internet_churn = pd.crosstab(
    df["InternetService"],
    df["Churn"],
    normalize="index"
)

internet_churn
```

```python
internet_churn.plot(kind="bar")

plt.title("Churn Rate by Internet Service")
plt.xlabel("Internet Service")
plt.ylabel("Proportion")
plt.xticks(rotation=0)
plt.show()
```

Customers using fiber optic internet show a higher churn rate than DSL users or customers without internet service.
